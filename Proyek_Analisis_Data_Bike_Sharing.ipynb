## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

### Assessing Data

#### Identifying ... problem

```python

# Memeriksa nilai yang tidak valid pada dataset

# Kolom kategori dan rentang nilai yang diperbolehkan
valid_categories = {
    "season": [1, 2, 3, 4],
    "yr": [0, 1],
    "mnth": list(range(1, 13)),
    "weekday": list(range(0, 7)),
    "holiday": [0, 1],
    "workingday": [0, 1],
    "weathersit": [1, 2, 3, 4]
}

# Fungsi pemeriksaan nilai tidak valid
def check_invalid_values(df, dataset_name):
    print(f"\nPemeriksaan {dataset_name}")

    for column, valid_values in valid_categories.items():
        if column in df.columns:
            invalid_count = (~df[column].isin(valid_values)).sum()
            print(f"{column}: {invalid_count} nilai tidak valid")

    # Memeriksa nilai numerik yang seharusnya tidak negatif
    non_negative_columns = [
        "instant", "casual", "registered", "cnt"
    ]

    for column in non_negative_columns:
        if column in df.columns:
            invalid_count = (df[column] < 0).sum()
            print(f"{column} negatif: {invalid_count}")

    # Memeriksa nilai normalisasi
    normalized_columns = ["temp", "atemp", "hum", "windspeed"]

    for column in normalized_columns:
        if column in df.columns:
            invalid_count = (
                ~df[column].between(0, 1)
            ).sum()
            print(f"{column} di luar rentang 0-1: {invalid_count}")

    # Pemeriksaan jam khusus dataset hour
    if "hr" in df.columns:
        invalid_hour = (~df["hr"].between(0, 23)).sum()
        print(f"Jam tidak valid: {invalid_hour}")


check_invalid_values(day_df, "DAY")
check_invalid_values(hour_df, "HOUR")
```

### Cleaning Data

### Cleaning Data

Tahap cleaning data dilakukan untuk memastikan dataset memiliki kualitas yang baik sebelum digunakan dalam proses eksplorasi dan analisis.

Pembersihan dilakukan dengan menghapus data duplikat, menangani missing values, memperbaiki nilai yang tidak sesuai dengan rentang yang diperbolehkan, mengubah tipe data tanggal, serta memeriksa konsistensi jumlah penyewaan.

#### Fixing ... problem

```python
# Membuat salinan dataset
day_clean = day_df.copy()
hour_clean = hour_df.copy()

# Menghapus data duplikat
day_clean.drop_duplicates(inplace=True)
hour_clean.drop_duplicates(inplace=True)

# Mengubah kolom tanggal menjadi datetime
day_clean["dteday"] = pd.to_datetime(
    day_clean["dteday"], errors="coerce"
)

hour_clean["dteday"] = pd.to_datetime(
    hour_clean["dteday"], errors="coerce"
)

# Menghapus baris dengan tanggal tidak valid
day_clean.dropna(subset=["dteday"], inplace=True)
hour_clean.dropna(subset=["dteday"], inplace=True)

print("Cleaning data selesai.")
print("Ukuran day_clean:", day_clean.shape)
print("Ukuran hour_clean:", hour_clean.shape)
```

```python

# Memeriksa kembali kualitas data setelah cleaning

print("MISSING VALUES DAY:")
print(day_clean.isnull().sum().sum())

print("\nMISSING VALUES HOUR:")
print(hour_clean.isnull().sum().sum())

print("\nDUPLICATE DAY:")
print(day_clean.duplicated().sum())

print("\nDUPLICATE HOUR:")
print(hour_clean.duplicated().sum())

print("\nKONSISTENSI JUMLAH PENYEWAAN DAY:")
print(
    (day_clean["cnt"] ==
     day_clean["casual"] + day_clean["registered"]).all()
)

print("\nKONSISTENSI JUMLAH PENYEWAAN HOUR:")
print(
    (hour_clean["cnt"] ==
     hour_clean["casual"] + hour_clean["registered"]).all()
)
```

**Insight:** (Opsional)

Setelah proses cleaning data, dilakukan pemeriksaan ulang untuk memastikan bahwa data yang digunakan sudah tidak memiliki missing values dan duplicate data. Selain itu, tipe data tanggal telah disesuaikan menjadi datetime dan konsistensi jumlah penyewaan diperiksa kembali. Dataset hasil pembersihan selanjutnya digunakan dalam proses Exploratory Data Analysis (EDA).


## Exploratory Data Analysis (EDA)

### Exploratory Data Analysis (EDA)

Pada tahap ini dilakukan eksplorasi terhadap dataset Bike Sharing untuk memahami karakteristik penyewaan sepeda berdasarkan waktu dan kondisi cuaca.

Eksplorasi dilakukan menggunakan proses grouping dan agregasi untuk memperoleh informasi mengenai rata-rata jumlah penyewaan, pola permintaan berdasarkan jam, serta perbedaan jumlah penyewaan pada setiap kondisi cuaca.

Hasil eksplorasi akan digunakan sebagai dasar dalam menjawab pertanyaan bisnis yang telah ditentukan.

### Explore ...

```python

# Menghitung rata-rata penyewaan berdasarkan jam
hourly_rentals = (
    hour_clean.groupby("hr")["cnt"]
    .mean()
    .reset_index()
)

hourly_rentals.columns = [
    "hour", "average_rentals"
]

display(hourly_rentals)

# Mencari jam dengan rata-rata penyewaan tertinggi
peak_hour = hourly_rentals.loc[
    hourly_rentals["average_rentals"].idxmax()
]

# Mencari jam dengan rata-rata penyewaan terendah
lowest_hour = hourly_rentals.loc[
    hourly_rentals["average_rentals"].idxmin()
]

print(
    f"Jam dengan rata-rata penyewaan tertinggi: "
    f"{int(peak_hour['hour'])}:00"
)

print(
    f"Rata-rata penyewaan: "
    f"{peak_hour['average_rentals']:.2f}"
)

print(
    f"\nJam dengan rata-rata penyewaan terendah: "
    f"{int(lowest_hour['hour'])}:00"
)

print(
    f"Rata-rata penyewaan: "
    f"{lowest_hour['average_rentals']:.2f}"
)
```

```python

# Membuat label kondisi cuaca
weather_labels = {
    1: "Clear / Partly Cloudy",
    2: "Mist / Cloudy",
    3: "Light Rain / Snow",
    4: "Heavy Rain / Snow"
}

day_clean["weather_label"] = (
    day_clean["weathersit"].map(weather_labels)
)

# Menghitung rata-rata penyewaan berdasarkan cuaca
weather_rentals = (
    day_clean.groupby("weather_label")["cnt"]
    .mean()
    .sort_values(ascending=False)
    .reset_index()
)

weather_rentals.columns = [
    "weather_condition", "average_rentals"
]

display(weather_rentals)

# Kondisi cuaca dengan penyewaan tertinggi dan terendah
best_weather = weather_rentals.iloc[0]
worst_weather = weather_rentals.iloc[-1]

print(
    "Kondisi cuaca dengan rata-rata penyewaan tertinggi:",
    best_weather["weather_condition"]
)

print(
    "Rata-rata penyewaan:",
    round(best_weather["average_rentals"], 2)
)

print(
    "\nKondisi cuaca dengan rata-rata penyewaan terendah:",
    worst_weather["weather_condition"]
)

print(
    "Rata-rata penyewaan:",
    round(worst_weather["average_rentals"], 2)
)
```

```python

# Label musim
season_labels = {
    1: "Spring",
    2: "Summer",
    3: "Fall",
    4: "Winter"
}

day_clean["season_label"] = (
    day_clean["season"].map(season_labels)
)

# Rata-rata penyewaan berdasarkan musim
season_rentals = (
    day_clean.groupby("season_label")["cnt"]
    .mean()
    .sort_values(ascending=False)
    .reset_index()
)

season_rentals.columns = [
    "season", "average_rentals"
]

display(season_rentals)
```

**Insight:** (Opsional)

1. **Pola Penyewaan Berdasarkan Jam (Hourly Rentals):**
   - Jam sibuk atau puncak penyewaan sepeda terjadi pada sore hari, tepatnya pukul **17:00** dengan rata-rata penyewaan mencapai sekitar **461.45** sepeda. Hal ini kemungkinan besar dipengaruhi oleh jam pulang kantor atau sekolah.
   - Tingkat penyewaan paling rendah terjadi pada dini hari pukul **04:00** dengan rata-rata hanya **6.35** sepeda.

2. **Pengaruh Kondisi Cuaca (Weather Condition):**
   - Penyewaan sepeda sangat sensitif terhadap cuaca. Rata-rata penyewaan tertinggi terjadi pada kondisi cuaca **Clear / Partly Cloudy** (Cerah/Berawan Sebagian) dengan rata-rata **4.876,79** sepeda per hari.
   - Sebaliknya, ketika kondisi cuaca memburuk seperti **Light Rain / Snow** (Hujan Ringan/Salju), minat masyarakat untuk bersepeda turun drastis hingga menyisakan rata-rata **1.803,29** sepeda per hari.

3. **Pengaruh Musim (Seasonal Rentals):**
   - Musim gugur (**Fall**) merupakan musim paling favorit bagi pelanggan untuk menyewa sepeda dengan rata-rata mencapai **5.644,30** penyewaan per hari, diikuti oleh musim panas (**Summer**) sebesar **4.992,33** penyewaan per hari.
   - Musim semi (**Spring**) mencatat jumlah penyewaan terendah dengan rata-rata hanya **2.604,13** sepeda per hari.

## Visualization & Explanatory Analysis

### Pertanyaan 1: Pola Penyewaan Sepeda Berdasarkan Jam

Visualisasi ini bertujuan untuk mengetahui pola rata-rata jumlah penyewaan sepeda pada setiap jam selama periode 2011–2012.

Data dikelompokkan berdasarkan jam, kemudian dihitung rata-rata jumlah penyewaan menggunakan atribut `cnt`. Visualisasi menggunakan grafik garis agar perubahan permintaan penyewaan sepanjang hari dapat terlihat dengan jelas.

### Pertanyaan 1:

```python

# Visualisasi rata-rata penyewaan berdasarkan jam

plt.figure(figsize=(14, 6))

sns.lineplot(
    data=hourly_rentals,
    x="hour",
    y="average_rentals",
    marker="o",
    linewidth=2.5
)

plt.title(
    "Average Bike Rentals by Hour (2011–2012)",
    fontsize=16,
    pad=15
)

plt.xlabel("Hour of the Day")
plt.ylabel("Average Number of Rentals")

plt.xticks(range(0, 24))
plt.grid(axis="y", alpha=0.3)

# Menandai jam tertinggi dan terendah
plt.annotate(
    f"Peak: {int(peak_hour['hour'])}:00",
    xy=(
        peak_hour["hour"],
        peak_hour["average_rentals"]
    ),
    xytext=(10, 15),
    textcoords="offset points",
    arrowprops=dict(arrowstyle="->")
)

plt.annotate(
    f"Lowest: {int(lowest_hour['hour'])}:00",
    xy=(
        lowest_hour["hour"],
        lowest_hour["average_rentals"]
    ),
    xytext=(10, -20),
    textcoords="offset points",
    arrowprops=dict(arrowstyle="->")
)

plt.tight_layout()
plt.show()
```

   - Tidak terdapat data penyewaan pada kategori Heavy Rain / Snow dalam dataset harian. Dengan demikian, kategori tersebut tidak dapat dibandingkan berdasarkan rata-rata penyewaan karena tidak memiliki observasi.

## Analisis Lanjutan (Opsional)

```python

# Analisis lanjutan 1: Jumlah penyewaan berdasarkan kelompok waktu

# Membuat salinan dataset
advanced_df = hour_clean.copy()

# Mengelompokkan jam menjadi beberapa periode
def categorize_time(hour):
    if 5 <= hour <= 10:
        return "Pagi"
    elif 11 <= hour <= 15:
        return "Siang"
    elif 16 <= hour <= 19:
        return "Sore"
    else:
        return "Malam"

advanced_df["time_period"] = advanced_df["hr"].apply(categorize_time)

# Menghitung rata-rata penyewaan setiap periode
time_analysis = (
    advanced_df.groupby("time_period", observed=False)["cnt"]
    .mean()
    .reindex(["Pagi", "Siang", "Sore", "Malam"])
    .reset_index()
)

print("Rata-rata penyewaan berdasarkan kelompok waktu:")
display(time_analysis)

# Menentukan periode dengan rata-rata penyewaan tertinggi dan terendah
highest_period = time_analysis.loc[time_analysis["cnt"].idxmax()]
lowest_period = time_analysis.loc[time_analysis["cnt"].idxmin()]

print(f"Periode tertinggi: {highest_period['time_period']} "
      f"({highest_period['cnt']:.2f} penyewaan)")
print(f"Periode terendah: {lowest_period['time_period']} "
      f"({lowest_period['cnt']:.2f} penyewaan)")


# Analisis lanjutan 2: Jumlah penyewaan berdasarkan kategori suhu

# Membuat kategori suhu berdasarkan nilai normalisasi
temp_bins = [0, 0.3, 0.6, 1.0]
temp_labels = ["Suhu Rendah", "Suhu Sedang", "Suhu Tinggi"]

advanced_df["temp_category"] = pd.cut(
    advanced_df["temp"],
    bins=temp_bins,
    labels=temp_labels,
    include_lowest=True
)

# Menghitung rata-rata penyewaan berdasarkan kategori suhu
temp_analysis = (
    advanced_df.groupby("temp_category", observed=False)["cnt"]
    .mean()
    .reset_index()
)

print("\nRata-rata penyewaan berdasarkan kategori suhu:")
display(temp_analysis)

# Menentukan kategori suhu dengan rata-rata tertinggi dan terendah
highest_temp = temp_analysis.loc[temp_analysis["cnt"].idxmax()]
lowest_temp = temp_analysis.loc[temp_analysis["cnt"].idxmin()]

print(f"Kategori tertinggi: {highest_temp['temp_category']} "
      f"({highest_temp['cnt']:.2f} penyewaan)")
print(f"Kategori terendah: {lowest_temp['temp_category']} "
      f"({lowest_temp['cnt']:.2f} penyewaan)")
```


**Insight:**

- Berdasarkan pengelompokan waktu, terdapat perbedaan rata-rata jumlah penyewaan sepeda pada periode pagi, siang, sore, dan malam. Periode dengan rata-rata penyewaan tertinggi adalah **Sore** (**377.63** penyewaan), sedangkan periode terendah adalah **Malam** (**83.73** penyewaan).

- Berdasarkan kategori suhu, rata-rata jumlah penyewaan tertinggi terdapat pada kategori **Suhu Tinggi** (**268.66** penyewaan), sedangkan rata-rata terendah terdapat pada kategori **Suhu Rendah** (**84.38** penyewaan).Hasil ini menunjukkan adanya pola bahwa rata-rata penyewaan sepeda cenderung lebih tinggi pada kategori suhu yang lebih tinggi dalam dataset yang dianalisis.
